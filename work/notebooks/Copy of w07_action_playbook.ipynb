## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*

## Ranked actions + reason codes

The model produces a ranked list of content pages for review.

Priority order:

1. High priority
   - Large decline in impressions
   - Poor average search position
   - Long time since last update

2. Medium priority
   - Moderate traffic decline
   - Stable rankings but decreasing CTR

3. Low priority
   - Stable search performance
   - No strong decline signals

Reason codes help explain why a page appears in the queue. These recommendations are intended to support human decisions rather than replace them.

## 2. Intended use and limits

*Who uses this, for what — and where it stops being valid.*

## Intended use and limits

This playbook is intended for SEO specialists, content teams, and website managers who need to prioritize content updates.

The recommendations are based on observed historical search signals and should be used as decision-support.

The model cannot explain why Google changed rankings, predict future rankings, or guarantee that updating a page will improve performance.

## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

## Human review + no-go list

Before acting on any recommendation, a reviewer should check:

- Whether the content is still accurate.
- Whether search intent has changed.
- Whether the page is strategically important.
- Whether technical issues affect performance.

The following actions should never be automated:

- Publishing content updates.
- Deleting pages.
- Changing business-critical information.
- Making final SEO decisions without human review.

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

df["is_declining"] = (
    df["trend_direction"]
      .str.lower()
      .eq("down")
      .astype(int)
)

review_queue = df[
    [
        "impressions_90d",
        "avg_position",
        "ctr",
        "days_since_last_update",
        "is_declining"
    ]
]

print("Review queue size:", len(review_queue))

review_queue.head()
```

## 4. Monitoring / retrain triggers

*What would tell you the recommendations went stale?*

## Monitoring and retraining

The model should be reviewed when:

- Search traffic patterns change significantly.
- Prediction accuracy decreases.
- New content types are added.
- Search engine behaviour changes over time.
- Fresh historical data becomes available.

Regular monitoring helps ensure that recommendations remain useful and representative of current search performance.

```python
print("Declining pages:",
      df["is_declining"].sum())

print("Declining rate:",
      round(df["is_declining"].mean(),3))
```

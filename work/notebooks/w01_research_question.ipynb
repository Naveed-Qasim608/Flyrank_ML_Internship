## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

# Load dataset
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Total pages:", len(df))

# Declining pages
df["is_declining"] = df["trend_direction"].str.lower().eq("down")

print("Declining pages:", df["is_declining"].sum())

print("Declining rate:", round(df["is_declining"].mean(), 3))

# Average impressions comparison
print(
    "Average impressions - declining pages:",
    round(df.loc[df["is_declining"], "impressions_90d"].mean(), 2)
)

print(
    "Average impressions - stable pages:",
    round(df.loc[~df["is_declining"], "impressions_90d"].mean(), 2)
)
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

## What this work can and cannot claim

### What I can say:
- **Observed:** I can identify patterns in available search and content signals, such as impressions, ranking changes, query coverage, and engagement metrics.
- **Directional:** The model can highlight pages that show signals associated with declining performance and may need attention.
- **Decision-support:** The output can help teams prioritize which pages to review or refresh first based on measurable signals.

### What I cannot say:
- This analysis does **not prove causation**. It cannot prove that a specific factor caused a page to decline.
- It does **not predict Google's ranking algorithm** or explain exactly why Google changes rankings.
- It does **not guarantee future traffic outcomes** after a content update.

The goal is to create an explainable decision-support tool that helps prioritize content reviews, not to claim causal explanations or predict search engine behavior.
